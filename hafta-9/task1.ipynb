Metni byte'lara çevir. Türkçe harfler içeren bir cümle seç (ç, ğ, ı, ö, ş, ü). Her harfin Unicode kod noktasını ve UTF-8 byte'larını yazdır. Hangi harfler 1 byte, hangileri 2 byte tutuyor? Kod noktalarını doğrudan vocab olarak kullanmak neden kötü bir fikir, kendi cümlelerinle yaz. (0:14:56 - 0:23:50)

```python
import pandas as pd
```

### Unicode kod noktaları
`ord()` bir karakterin Unicode kod noktasını (tam sayı) verir.

```python
print(ord("h"))
print(ord("ş"))
print(ord("😊"))
```

```python
cumle = "Öğle güneşinde çocuklar ışıl ışıl şarkı söylüyordu."
print([ord(x) for x in cumle])
```

### UTF-8 / UTF-16 / UTF-32

```python
print("utf-8 :", list(cumle.encode("utf-8")))
print("utf-16:", list(cumle.encode("utf-16"))[:30], "...")
print("utf-32:", list(cumle.encode("utf-32"))[:30], "...")
print()
print("karakter sayısı   :", len(cumle))
print("utf-8 byte sayısı :", len(cumle.encode("utf-8")))
print("utf-16 byte sayısı:", len(cumle.encode("utf-16")))
print("utf-32 byte sayısı:", len(cumle.encode("utf-32")))
```

### Her harfin kod noktası ve UTF-8 byte'ları

```python
for ch in cumle:
    b = ch.encode("utf-8")
    print(f"{ch!r:5} kod noktası: {ord(ch):5}  (U+{ord(ch):04X})  utf-8: {list(b)!s:12} {len(b)} byte")
```

### 1 byte / 2 byte harfler

```python
harfler = sorted(set(ch for ch in cumle if ch.isalpha()))
df = pd.DataFrame({
    "harf": harfler,
    "kod noktası": [ord(h) for h in harfler],
    "hex": [f"U+{ord(h):04X}" for h in harfler],
    "utf-8 byte'ları": [list(h.encode("utf-8")) for h in harfler],
    "byte sayısı": [len(h.encode("utf-8")) for h in harfler],
})
df.sort_values(["byte sayısı", "kod noktası"]).reset_index(drop=True)
```

```python
print("1 byte:", [h for h in harfler if len(h.encode("utf-8")) == 1])
print("2 byte:", [h for h in harfler if len(h.encode("utf-8")) == 2])
```

```python
# Türk alfabesine özgü tüm harfler (büyük + küçük)
for ch in "çğıöşüÇĞİÖŞÜ":
    b = ch.encode("utf-8")
    print(f"{ch}  U+{ord(ch):04X}  {list(b)}  {len(b)} byte")
```

### Yorum

**Hangi harfler kaç byte?**
İngiliz alfabesinde de bulunan harfler (a, e, k, l, r, s, u ... ve boşluk, nokta) ASCII aralığında (kod noktası < 128) olduğu için UTF-8'de **1 byte** tutuyor. Türkçeye özgü **ç, ğ, ı, ö, ş, ü** (ve büyükleri Ç, Ğ, İ, Ö, Ş, Ü) ise 128-2047 aralığında olduğu için **2 byte** tutuyor. Örneğin `ş` = U+015F → `[197, 159]`, `ı` = U+0131 → `[196, 177]`. Yani aynı uzunluktaki Türkçe bir cümle, İngilizce bir cümleden daha fazla byte'a dönüşüyor (cümlemde 51 karakter → 66 byte, çünkü 15 tane 2 byte'lık harf var). Emoji gibi karakterler ise 4 byte'a kadar çıkıyor.

UTF-16 ve UTF-32'de ise İngilizce harfler bile 2 / 4 byte tutuyor ve araya bir sürü `0` giriyor; bu yüzden israf çok fazla, UTF-8 en mantıklı seçim.

**Kod noktalarını doğrudan vocab olarak kullanmak neden kötü?**
1. **Vocab çok büyük olur:** Unicode'da ~150.000 karakter tanımlı. Her biri bir token olursa embedding tablosu ve son katmandaki (lm_head) çıktı boyutu devasa olur, ama bu karakterlerin çok büyük kısmı eğitim verisinde neredeyse hiç geçmez; o satırlar düzgün eğitilemez, boşa parametre olur.
2. **Standart sabit değil:** Unicode her yıl yeni karakterler (özellikle emoji) ekliyor. Vocab'ı kod noktalarına bağlarsam, standart değiştikçe tokenizer'ı ve modeli değiştirmem gerekir.
3. **Hiç sıkıştırma yok:** Her karakter bir token olduğu için diziler çok uzun olur. Transformer'ın context penceresi (block_size) sınırlı, attention maliyeti de dizi uzunluğuyla karesel artıyor; karakter karakter gidince model aynı pencerede çok az metin görür.

Byte'lara (UTF-8) inince vocab 256'ya düşüyor ve her metni temsil edebiliyoruz ama bu sefer diziler daha da uzuyor. Bu yüzden ham byte'ları alıp **BPE** ile sık geçen byte ikililerini birleştirerek hem vocab'ı kontrol edilebilir boyutta tutmak hem de dizileri kısaltmak gerekiyor.
