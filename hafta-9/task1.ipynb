Metni byte'lara çevir. Türkçe harfler içeren bir cümle seç (ç, ğ, ı, ö, ş, ü). Her harfin Unicode kod noktasını ve UTF-8 byte'larını yazdır. Hangi harfler 1 byte, hangileri 2 byte tutuyor? Kod noktalarını doğrudan vocab olarak kullanmak neden kötü bir fikir, kendi cümlelerinle yaz. (0:14:56 - 0:23:50)

### Unicode kod noktaları
`ord()` bir karakterin Unicode kod noktasını (tam sayı) verir.

```python
print(ord("h"))
print(ord("ş"))
print(ord("😊"))
```

```python
cumle = "Öğle güneşinde çocuklar ışıl ışıl şarkı söylüyordu."
print("Harflerin Unicode Tablosundaki Numaraları")
print([ord(x) for x in cumle])
```

### UTF-8 / UTF-16 / UTF-32

```python
print("utf-8 :", list(cumle.encode("utf-8")))
print("utf-16:", list(cumle.encode("utf-16"))[:30])
print("utf-32:", list(cumle.encode("utf-32"))[:30])

print("karakter sayısı:", len(cumle))
print("utf-8 byte sayısı :", len(cumle.encode("utf-8")))
print("utf-16 byte sayısı:", len(cumle.encode("utf-16")))
print("utf-32 byte sayısı:", len(cumle.encode("utf-32")))
```

### Her harfin kod noktası ve UTF-8 byteları

```python
for ch in cumle:
    b = ch.encode("utf-8")
    print(f"{ch!r:5} karakterinin Unicode tablosundaki numarası: {ord(ch):5}  (U+{ord(ch):04X})  utf-8: {list(b)!s:12} {len(b)} byte")
```

### 1 byte / 2 byte harfler

```python
import pandas as pd
```

```python
harfler = sorted(set(ch for ch in cumle if ch.isalpha()))
df = pd.DataFrame({
    "harf": harfler,
    "Unicode": [ord(h) for h in harfler],
    "hex": [f"U+{ord(h):04X}" for h in harfler],
    "utf-8 byteları": [list(h.encode("utf-8")) for h in harfler],
    "byte sayısı": [len(h.encode("utf-8")) for h in harfler],
})
df.sort_values(["byte sayısı", "Unicode"]).reset_index(drop=True)
```

ASCII tablosunda 128 karakter vardır ve bunlar ingiliz alfabesi ve bazı özel karakterlerdir. Türkçe karakterler ise unicode sıralaması 128den büyük olduğu için tek byte yetmiyor bundan dolayı iki bytelık yer kaplıyorlar.

Unicode her yıl yenileniyor yeni karakterler ekleniyor bu sürdürülebilir bir şey değildir her yıl modeli ve tokenizerı değiştirmek zorunda kalacaktık.
